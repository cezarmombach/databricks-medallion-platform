```python
# Method: pytest.main() called from a notebook
# Objective: Run the unit tests in /tests and fail loudly if any test fails.
import sys
import pytest

sys.dont_write_bytecode = True  # avoid stale cached bytecode after editing files

# Forget project modules imported earlier, so edits to src/ are picked up on rerun
for name in [m for m in sys.modules if m == "src" or m.startswith("src.")]:
    del sys.modules[name]

# "../tests" because this notebook lives inside notebooks/ and tests/ is a sibling folder
exit_code = pytest.main(["-v", "-p", "no:cacheprovider", "../tests"])

# exit code 0 means all passed; anything else means at least one test failed
assert exit_code == 0, "Some unit tests failed"
```
